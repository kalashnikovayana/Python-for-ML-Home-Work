````python
import os
from datetime import datetime

from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.tools import tool
from langchain.agents import create_react_agent, AgentExecutor

# ── DuckDuckGo (без ключів) ───────────────────────────────────────────────────
from langchain_community.tools import DuckDuckGoSearchRun
web_search = DuckDuckGoSearchRun(name="duckduckgo_latest")

# ── arXiv (через офіційний пакет, а не LangChain-обгортку) ───────────────────
import arxiv

client = arxiv.Client()  # окремий клієнт, щоб швидко витягувати результати

@tool("arxiv_search")
def arxiv_search(query: str, max_results: int = 5):
    """Пошук у arXiv. Повертає список словників: title, authors, summary, published, url."""
    search = arxiv.Search(
        query=query,
        max_results=max_results,
        sort_by=arxiv.SortCriterion.SubmittedDate,
        sort_order=arxiv.SortOrder.Descending,
    )
    results = client.results(search)
    out = []
    for r in results:
        out.append({
            "title": r.title,
            "authors": [a.name for a in r.authors],
            "summary": r.summary,
            "published": r.published.strftime("%Y-%m-%d") if r.published else None,
            "url": r.entry_id,
        })
    return out
````

````python
REACT_PROMPT = PromptTemplate.from_template("""
Ви агент-дослідник. У вас є доступ до інструментів:
{tools}

Використовуйте лише ці інструменти: {tool_names}.

Формат:
Thought:
Action: один із [{tool_names}]
Action Input:
Observation:
(повторюйте за потреби)

Коли достатньо даних, дайте відповідь:
Final Answer: поверніть JSON-масив рівно з 5 об'єктів:
- title (рядок)
- authors (список рядків)
- summary (2–4 речення українською)
- published (YYYY-MM-DD)
- source_url (url)

Підказки:
- Спочатку скористайтесь duckduckgo_latest, щоб знайти свіжі згадки/назви за останній місяць (за потреби додавайте "site:arxiv.org").
- Потім викликайте arxiv_search для коректних метаданих.
- Якщо результатів >5 — залишайте найновіші.

Питання користувача:
{input}

{agent_scratchpad}
""")
````

````python
def build_agent(model="gpt-4o-mini"):
    llm = ChatOpenAI(model=model, temperature=0.2)  # потрібен OPENAI_API_KEY
    tools = [web_search, arxiv_search]
    agent = create_react_agent(llm, tools, prompt=REACT_PROMPT)
    return AgentExecutor(agent=agent, tools=tools, verbose=True, handle_parsing_errors=True)

agent = build_agent()

query = ("Знайди 5 останніх публікацій на тему штучного інтелекту (Artificial Intelligence) "
         "за останній місяць і поверни JSON з title, authors, summary, published, source_url.")
res = agent.invoke({"input": query})
print(res["output"])
````

````python
# --- imports ---
import os, requests, arxiv, numpy as np
from typing import Dict, Union
from pydantic import BaseModel, Field

from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.tools import tool, Tool
from langchain.agents import create_react_agent, AgentExecutor

from ddgs import DDGS
from bs4 import BeautifulSoup
import json, re

# ── Пошук DuckDuckGo ─────────────────────────────────────────────────────────
from urllib.parse import urlparse
from ddgs import DDGS
from langchain_core.tools import tool

WHITELIST = {
    "fao.org", "oecd.org", "usda.gov", "imf.org",
    "conab.gov.br", "inmet.gov.br", "citrusbr.com",
    "citrusbr.com.br"
}
BLOCKED_HOSTS = {
    "bing.com", "news.yahoo.com", "s.click", "doubleclick.net",
    "googleadservices.com", "ads.", "sundale", "netstock"
}

def _good(url: str) -> bool:
    try:
        host = urlparse(url).netloc.lower()
        # зрізаємо піддомени: *.usda.gov => usda.gov
        parts = host.split(".")
        root = ".".join(parts[-2:]) if len(parts) >= 2 else host
        if any(b in host for b in BLOCKED_HOSTS):
            return False
        return (root in WHITELIST) or any(host.endswith(w) for w in WHITELIST)
    except Exception:
        return False

@tool("duckduckgo_latest", return_direct=False)
def duckduckgo_latest(query: str, max_results: int = 20, timelimit: str = "m"):
    """
    Пошук DuckDuckGo з фільтром свіжості ('d','w','m','y').
    Повертає лише результати з білих доменів (FAO/OECD/USDA/IMF/CONAB/INMET/CITRUSBR).
    Формат: [{title, href, body}, ...]
    """
    out, seen = [], set()
    with DDGS() as ddgs:
        try:
            gen = ddgs.text(query, max_results=max_results, timelimit=timelimit)
        except TypeError:  # старі версії
            gen = ddgs.text(keywords=query, max_results=max_results, timelimit=timelimit)

        for r in gen:
            href = r.get("href")
            if not href or not _good(href):
                continue
            key = urlparse(href).netloc + urlparse(href).path
            if key in seen:
                continue
            seen.add(key)
            out.append({"title": r.get("title"), "href": href, "body": r.get("body")})
    return out

# ── Завантаження сторінок ────────────────────────────────────────────────────
import time, requests
from langchain_core.tools import tool

@tool("requests_get", return_direct=False)
def fetch_url(url: str, max_chars: int = 12000) -> str:
    """
    GET з повторними спробами: таймаути (connect, read) ростуть 6→15→35s.
    Повертає перші max_chars символів тексту/HTML.
    """
    headers = {
        "User-Agent": "Mozilla/5.0 (compatible; LangChain-Agent/1.0)"
    }
    timeouts = [(6.0, 15.0), (10.0, 25.0), (15.0, 35.0)]  # (connect, read)
    last_err = None
    for conn_t, read_t in timeouts:
        try:
            r = requests.get(url, headers=headers, timeout=(conn_t, read_t), allow_redirects=True)
            r.raise_for_status()
            return r.text[:max_chars]
        except (requests.ReadTimeout, requests.ConnectionError) as e:
            last_err = e
            time.sleep(0.7)  # невелика пауза і ще спроба
        except Exception as e:
            last_err = e
            break
    raise RuntimeError(f"requests_get failed after retries: {last_err}")

# ── Очищення HTML → текст ────────────────────────────────────────────────────
@tool("extract_text", return_direct=False)
def extract_text(html: str, max_chars: int = 5000) -> str:
    """Обрізає HTML до plain text (перші max_chars символів)."""
    soup = BeautifulSoup(html, "lxml")
    text = soup.get_text(separator="\n")
    return text[:max_chars]

# ── Python REPL (fallback) ───────────────────────────────────────────────────
try:
    from langchain_experimental.tools.python.tool import PythonREPLTool
    py_tool = PythonREPLTool()
except Exception:
    from langchain_experimental.utilities import PythonREPL
    py_tool = Tool(
        name="python_repl",
        func=PythonREPL().run,
        description="Виконує невеликі обчислення в Python. Без I/O."
    )

# ── Базовий прогноз (структуровані аргументи) ────────────────────────────────
class BaselineArgs(BaseModel):
    series_kv: Dict[str, Union[int, float]] = Field(
        ...,
        description="Словник рік→тонни, напр.: {'2021':200,'2022':190,'2023':210,'2024':220}"
    )
    target_year: int = Field(2025, description="Рік прогнозу")

@tool("baseline_forecast", return_direct=False)
def baseline_forecast(payload: Union[str, Dict[str, Union[int, float]]]) -> float:
    """
    Гнучкий тул прогнозу.
    Приймає:
    1) JSON-рядок: {"series_kv":{"2021":200,"2022":190,"2023":210,"2024":220},"target_year":2025}
    2) Рядок типу: {"2021":200,"2022":190,"2023":210,"2024":220}, 2025
    3) Просто словник: {"2021":200,"2022":190,"2023":210,"2024":220}  (цільовий рік = 2025)
    Повертає baseline-прогноз (лінійна регресія), округлено до 0.1.
    """
    series_kv: Dict[str, float] = {}
    target_year = 2025

    def _fit_and_pred(sk: Dict[str, Union[int, float]], ty: int) -> float:
        years = np.array(sorted(int(k) for k in sk.keys()), dtype=float)
        vals  = np.array([float(sk[str(int(y))]) for y in years], dtype=float)
        coef  = np.polyfit(years, vals, 1)
        pred  = np.polyval(coef, ty)
        return float(round(pred, 1))

    # Випадок 1: прийшов dict
    if isinstance(payload, dict):
        if "series_kv" in payload:
            series_kv = {str(k): float(v) for k, v in payload["series_kv"].items()}
            target_year = int(payload.get("target_year", 2025))
        else:
            series_kv = {str(k): float(v) for k, v in payload.items()}
        return _fit_and_pred(series_kv, target_year)

    # Випадок 2: прийшов рядок
    s = str(payload).strip()

    # 2a) JSON-рядок зі schema
    try:
        obj = json.loads(s.replace("'", '"'))
        if isinstance(obj, dict) and "series_kv" in obj:
            series_kv = {str(k): float(v) for k, v in obj["series_kv"].items()}
            target_year = int(obj.get("target_year", 2025))
            return _fit_and_pred(series_kv, target_year)
    except Exception:
        pass

    # 2b) Рядок типу: {...}, 2025
    m = re.match(r"\s*(\{.*\})\s*,\s*(\d{4})\s*$", s)
    if m:
        left, year = m.group(1), m.group(2)
        try:
            series_kv = {str(k): float(v) for k, v in json.loads(left.replace("'", '"')).items()}
            return _fit_and_pred(series_kv, int(year))
        except Exception:
            pass

    # 2c) Просто {...}
    try:
        series_kv = {str(k): float(v) for k, v in json.loads(s.replace("'", '"')).items()}
        return _fit_and_pred(series_kv, target_year)
    except Exception:
        raise ValueError(
            "baseline_forecast: не вдалося розпізнати вхід. "
            "Очікуйте або JSON зі schema, або '{...}, 2025', або dict."
        )

# ── arXiv (офіційний пакет) ─────────────────────────────────────────────────
arxiv_client = arxiv.Client()

@tool("arxiv_search")
def arxiv_search(query: str, max_results: int = 5):
    """Пошук у arXiv. Повертає title, authors, summary, published, url."""
    search = arxiv.Search(
        query=query, max_results=max_results,
        sort_by=arxiv.SortCriterion.SubmittedDate,
        sort_order=arxiv.SortOrder.Descending,
    )
    out = []
    for r in arxiv_client.results(search):
        out.append({
            "title": r.title,
            "authors": [a.name for a in r.authors],
            "summary": r.summary,
            "published": r.published.strftime("%Y-%m-%d") if r.published else None,
            "url": r.entry_id,
        })
    return out
````

````python
REACT_PROMPT = PromptTemplate.from_template("""
Ви бізнес-аналітик-агент. Інструменти:
{tools}

Використовуйте лише: {tool_names}.

Формат:
Thought:
Action: один із [{tool_names}]
Action Input:
Observation:
(повторюйте)

Завдання:
1) Витягніть часовий ряд експорту з тексту користувача (рік→тонни).
2) Обчисліть базовий прогноз на 2025 через baseline_forecast.
3) Зберіть ФАКТИ для коригування на погоду в Бразилії та глобальний попит:
   - Використовуйте duckduckgo_latest з фразами типу:
     "Brazil orange harvest 2025 forecast site:conab.gov.br OR site:inmet.gov.br",
     "orange juice demand 2025 outlook site:fao.org OR site:oecd.org OR site:usda.gov",
     "global CPI 2025 forecast site:imf.org WEO".
   - Потім відкрийте ключові URL через requests_get і очистьте текст через extract_text.
   - Витягніть конкретні факти (дати, числа, прогнози).
4) Побудуйте коригування:
   adjusted = baseline * (1 + weather_factor + demand_inflation_factor)
   Де кожен фактор ∈ [-0.15; +0.15] і має бути обґрунтований знайденим фактом.
5) Уникайте фейкових джерел. Заборонені домени: example.com, блог-агрегатори без даних.
   Прийнятні: *.gov.br, fao.org, oecd.org, imf.org, usda.gov, citrusbr.com.br, conab.gov.br.
6) Фінал (лише JSON-подібний блок):
Final Answer:
- "baseline_forecast_tons_2025": число
- "adjusted_forecast_tons_2025": число
- "assumptions": 3–6 маркерів із короткими цитатами/датами
- "method": 1–2 речення (що рахували і як коригували)
- "sources": список 3–6 ПРАВИЛЬНИХ URL (https://...), без дублів

Питання користувача:
{input}

{agent_scratchpad}
""")
````

````python
def build_agent(model="gpt-4o-mini"):
    llm = ChatOpenAI(model=model, temperature=0.2)

    tools = [
        duckduckgo_latest,
        fetch_url,
        extract_text,
        baseline_forecast,
        arxiv_search,
        py_tool
    ]

    agent = create_react_agent(llm, tools, prompt=REACT_PROMPT)
    return AgentExecutor(agent=agent, tools=tools, verbose=True, handle_parsing_errors=True)
````
